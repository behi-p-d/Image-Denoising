```python
import sys

import os


# Add project root to Python path
project_root = r"your_path\Project"
sys.path.insert(0, project_root)

# Add the models directory (where shallow_autoencoder is)
sys.path.append(r"your_path\Project\models")

# Add the project root directory (where config is)
sys.path.append(r"your_path\Project\utils")

sys.path.append(r"your_path\Project\dataset")

from config import *

from dataset import BSD500Dataset, create_dataloaders
from experiment import run_experiment

from shallow_autoencoder import ShallowAutoencoder
from deep_autoencoder import DeepAutoencoder
from residual_autoencoder import ResidualAutoencoder

from losses import HybridLoss, MSELoss

import torch

set_seed()

print(f"Device: {DEVICE}")
```

```python
train_dataset = BSD500Dataset(
    image_dir=TRAIN_DIR,
    patch_size=PATCH_SIZE,
    patches_per_image=PATCHES_PER_IMAGE,
    add_noise=True
)

val_dataset = BSD500Dataset(
    image_dir=VAL_DIR,
    patch_size=PATCH_SIZE,
    patches_per_image=PATCHES_PER_IMAGE,
    add_noise=True
)

test_dataset = BSD500Dataset(
    image_dir=TEST_DIR,
    patch_size=PATCH_SIZE,
    patches_per_image=PATCHES_PER_IMAGE,
    add_noise=True
)
```

```python
train_loader, val_loader, test_loader = create_dataloaders(
    train_dataset,
    val_dataset,
    test_dataset,
    batch_size=BATCH_SIZE,
    num_workers=NUM_WORKERS,
)
```

```python
shallow_results = run_experiment(
    model_class=ShallowAutoencoder,
    criterion=MSELoss(),
    train_loader=train_loader,
    val_loader=val_loader,
    epochs=EPOCHS,
    model_name="shallow",
    loss_name="MSE",
)
print(f"Best Validation Loss: {shallow_results['best_val_loss']:.6f}")
```

```python
deep_results = run_experiment(
    model_class=DeepAutoencoder,
    criterion=MSELoss(),
    train_loader=train_loader,
    val_loader=val_loader,
    epochs=EPOCHS,
    model_name="deep",
    loss_name="MSE",
)
print(f"Best Validation Loss: {deep_results['best_val_loss']:.6f}")
```

```python
residual_results = run_experiment(
    model_class=ResidualAutoencoder,
    criterion=MSELoss(),
    train_loader=train_loader,
    val_loader=val_loader,
    epochs=EPOCHS,
    model_name="residual",
    loss_name="MSE",
)
print(f"Best Validation Loss: {residual_results['best_val_loss']:.6f}")
```

```python
results = [
    shallow_results,
    deep_results,
    residual_results,
]

for result in results:
    print(
    f"{result['model_name']:10s}"
    f"Validation Loss: {result['best_val_loss']:.6f}"    
    )
```

```python
residual_results = run_experiment(
    model_class=ResidualAutoencoder,
    criterion=HybridLoss(),
    train_loader=train_loader,
    val_loader=val_loader,
    epochs=EPOCHS,
    model_name="residual",
    loss_name="hybrid",
)
print(f"Best Validation Loss: {residual_results['best_val_loss']:.6f}")
```

```python
best_result = min(
    results,
    key=lambda x: x["best_val_loss"],
)

print("\nBest Architecture")
print(f"Model: {best_result['model_name']}")
print(f"Validation loss: {best_result['best_val_loss']:.6f}")
```

```python
best_hybrid_results= run_experiment(
    model_class=ResidualAutoencoder,
    criterion=HybridLoss(),
    train_loader=train_loader,
    val_loader=val_loader,
    epochs=EPOCHS,
    model_name="residual",
    loss_name="hybrid",
)

print(f"Best Validation Loss: {best_hybrid_results['best_val_loss']:.6f}")
```
