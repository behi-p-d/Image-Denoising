```python
import matplotlib.pyplot as plt
from noise import add_gaussian_noise, add_salt_pepper_noise, add_speckle_noise

# Get image
image = dataset[0]

# Add different noises
gaussian_noisy = add_gaussian_noise(image)
sp_noisy = add_salt_pepper_noise(image, amount_range=(0.01, 0.05))
speckle_noisy = add_speckle_noise(image, std_range=(0.05, 0.15))

# Visualize all four
fig, axes = plt.subplots(1, 4, figsize=(20, 5))

# Original
axes[0].imshow(image.permute(1, 2, 0))
axes[0].set_title("Original")
axes[0].axis("off")

# Gaussian Noise
axes[1].imshow(gaussian_noisy.permute(1, 2, 0))
axes[1].set_title("Gaussian Noise")
axes[1].axis("off")

# Salt & Pepper Noise
axes[2].imshow(sp_noisy.permute(1, 2, 0))
axes[2].set_title("Salt & Pepper Noise")
axes[2].axis("off")

# Speckle Noise
axes[3].imshow(speckle_noisy.permute(1, 2, 0))
axes[3].set_title("Speckle Noise")
axes[3].axis("off")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from noise import add_gaussian_noise, add_salt_pepper_noise, add_speckle_noise, add_mixed_noise

# Get image
image = dataset[18]

# Add different noises
gaussian_noisy = add_gaussian_noise(image)
sp_noisy = add_salt_pepper_noise(image, amount_range=(0.01, 0.05))
speckle_noisy = add_speckle_noise(image, std_range=(0.15, 0.35))
mixed_noisy = add_mixed_noise(image)

# Visualize all five
fig, axes = plt.subplots(1, 5, figsize=(25, 5))

# Original
axes[0].imshow(image.permute(1, 2, 0))
axes[0].set_title("Original")
axes[0].axis("off")

# Gaussian Noise
axes[1].imshow(gaussian_noisy.permute(1, 2, 0))
axes[1].set_title("Gaussian Noise")
axes[1].axis("off")

# Salt & Pepper Noise
axes[2].imshow(sp_noisy.permute(1, 2, 0))
axes[2].set_title("Salt & Pepper Noise")
axes[2].axis("off")

# Speckle Noise
axes[3].imshow(speckle_noisy.permute(1, 2, 0))
axes[3].set_title("Speckle Noise")
axes[3].axis("off")

# Mixed Noise
axes[4].imshow(mixed_noisy.permute(1, 2, 0))
axes[4].set_title("Mixed Noise")
axes[4].axis("off")

plt.tight_layout()
plt.show()
```
